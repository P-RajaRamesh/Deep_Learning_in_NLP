```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
### sentences
sent=[  'the glass of milk',
     'the glass of juice',
     'the cup of tea',
    'I am a good boy',
     'I am a good developer',
     'understand the meaning of words',
     'your videos are good',]
```

```python
# define vocabulary size -- for index size
voc_size=10000
```

```python
# one hot encoding
one_hot_repr=[one_hot(words,voc_size) for words in sent]
one_hot_repr

# around 10000 indexes, words are distributed.
# below output is the occurence of word at that index is 1 and remaning places for 10000 index all are zeros
```

```python
# word embedding representation
from tensorflow.keras.layers import Embedding
# from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
sent_length=8
# make every sentences length equal in size. So we do pre_pading
embedded_docs=pad_sequences(one_hot_repr,padding='pre',maxlen=sent_length)
embedded_docs
```

```python
# feature representation
dim=10
```

```python
model=Sequential()
model.add(Embedding(voc_size, dim, input_length=sent_length))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
model.predict(embedded_docs)
```
